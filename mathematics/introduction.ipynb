```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris_data = load_iris(as_frame=True)
X = iris_data.data
display(iris_data)
```

```python
sns.set_theme(style="darkgrid")

# Plot histograms for each feature in a 1x4 grid with clean custom colors
fig1, axes1 = plt.subplots(nrows=1, ncols=4, figsize=(20, 4.5))

# Define clean custom colors
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]

# Plot each histogram into its respective subplot slot
sns.histplot(data=X, x="sepal length (cm)", bins=15, color=colors[0], ax=axes1[0])
axes1[0].set_title("Sepal Length Histogram", fontweight="bold")

sns.histplot(data=X, x="sepal width (cm)", bins=15, color=colors[1], ax=axes1[1])
axes1[1].set_title("Sepal Width Histogram", fontweight="bold")

sns.histplot(data=X, x="petal length (cm)", bins=15, color=colors[2], ax=axes1[2])
axes1[2].set_title("Petal Length Histogram", fontweight="bold")

sns.histplot(data=X, x="petal width (cm)", bins=15, color=colors[3], ax=axes1[3])
axes1[3].set_title("Petal Width Histogram", fontweight="bold")

# Adjust spaces neatly and render the first figure
plt.tight_layout()
plt.show()

# Create the second figure for KDE Curves (1 row, 4 columns)
fig2, axes2 = plt.subplots(nrows=1, ncols=4, figsize=(20, 4.5))

# Plot each KDE curve into its respective subplot slot
sns.kdeplot(data=X, x="sepal length (cm)", fill=True, color=colors[0], ax=axes2[0])
axes2[0].set_title("Sepal Length KDE Curve", fontweight="bold")

sns.kdeplot(data=X, x="sepal width (cm)", fill=True, color=colors[1], ax=axes2[1])
axes2[1].set_title("Sepal Width KDE Curve", fontweight="bold")

sns.kdeplot(data=X, x="petal length (cm)", fill=True, color=colors[2], ax=axes2[2])
axes2[2].set_title("Petal Length KDE Curve", fontweight="bold")

sns.kdeplot(data=X, x="petal width (cm)", fill=True, color=colors[3], ax=axes2[3])
axes2[3].set_title("Petal Width KDE Curve", fontweight="bold")

# Adjust spaces neatly and render the second figure
plt.tight_layout()
plt.show()

```
