```python
from pandas import pandas as pd
from IPython.display import display
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
 
diabetes_frame = load_diabetes(as_frame=True)
X = diabetes_frame.data[['age']]  # Use double brackets to keep 2D shape
y = diabetes_frame.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, shuffle=False)
display((X_train.shape, X_test.shape, y_train.shape, y_test.shape))
```

## Training Model

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression().fit(X_train, y_train)
print(f"Coefficients: {regressor.coef_})")
print(f"Intercept: {regressor.intercept_}")
print(f"Rank of the coefficient matrix: {regressor.rank_}")
```

Model Attributes

## Model Evaluation

```python
from sklearn.metrics import mean_squared_error, r2_score

y_pred = regressor.predict(X_test)


print(f"Mean squared error: {mean_squared_error(y_test, y_pred):.2f}")
print(f"R2 score: {r2_score(y_test, y_pred):.2f}")
```

## Plotting the results

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(ncols=2, figsize=(12, 5), sharex=True, sharey=True)

print(type(y_train))
print(y_train.head())

ax[0].scatter(X_train, y_train, color='blue', label='Training data')
ax[0].plot(X_train, regressor.predict(X_train), color='red', label='Regression line')
ax[0].set_title('Training Data and Regression Line')
ax[0].set_xlabel('Age')
ax[0].set_ylabel('Diabetes Progression')
ax[0].legend()

ax[1].scatter(X_test, y_test, color='green', label='Test data')
ax[1].plot(X_test, y_pred, color='red', label='Regression line')
ax[1].set_title('Test Data and Regression Line')
ax[1].set_xlabel('Age')

ax[1].legend()

fig.suptitle('Linear Regression on Diabetes Dataset')

plt.show()
```
