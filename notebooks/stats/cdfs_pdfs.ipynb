```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from IPython.display import display
```

```python
X = np.linspace(0,5, 1001)

# for x in X:
#     print(x)
# plt.plot(X)


X = np.linspace(0,5, 1001)

p1 = stats.lognorm.pdf(X, 1)

c1 = stats.lognorm.cdf(X, 1)

p2 = stats.lognorm.pdf(X, 0.1)
c2 = stats.lognorm.cdf(X, 0.1)

display(f"PDFs: {p1} {p2}")
display(f"CDFs: {c1} {c2}")



```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].plot(X, p1, label='PDF (s=1)', color='blue')
ax[0].plot(X, p2, label='PDF (s=0.1)', color='orange')
ax[0].set_title('Log-Normal PDFs')
ax[0].set_xlabel('x')
ax[0].set_ylabel('Density')
ax[0].legend()

ax[1].plot(X, c1, label='CDF (s=1)', color='blue')
ax[1].plot(X, c2, label='CDF (s=0.1)', color='orange')
ax[1].set_title('Log-Normal CDFs')
ax[1].set_xlabel('x')
ax[1].set_ylabel('Cumulative Probability')
ax[1].legend()  

plt.show()
```
