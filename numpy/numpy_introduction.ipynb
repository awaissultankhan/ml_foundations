### Numpy Arrays

```python
X = np.array([90.1, 85.5, 92.3])
print(X)

print(type(X))
print(X.shape)
print(X.dtype)
```

```python
print(np.ones(shape=10))
print(np.zeros(shape=10))
print(np.random.rand(10))
```

### Numpy Arrays As Vectors

```python
v_1 = np.array([1, 2, 3])
v_2 = np.array([4, 5, 6])
v_3 = v_1 + v_2 # element-wise addition
print(v_3)
```

```python
v_4 = 10 * v_1  # scalar multiplication
print(v_4)
```

```python
v_5 = v_1 * v_2 # element-wise multiplication
print(v_5)
```

```python
print(np.zeros(shape=(2, 3)) + 1)
```

```python
from random import random as random_py
def builtin_random_single(n_runs):
    for _ in range(n_runs):
        random_py()

n_runs = 1000000
%prun builtin_random_single(n_runs)
```
